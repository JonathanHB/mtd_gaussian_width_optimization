```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#system
class double_well_1:
    def __init__(self, A,B):
        self.A = A
        self.B = B

    def f1(self, x):
        return self.A * (x**4 - self.B*x**2)

    def stdev(self):
        return np.sqrt(self.B/2)/2

    def well_volume(self,x):
        volume = 0
        for Fi in self.f1(x):
            if Fi<0:
                volume -= Fi
        return volume
```

```python
dw1 = double_well_1(1,5)

padfactor = 4
n_bins = 300
x = np.linspace(-padfactor*dw1.stdev(), padfactor*dw1.stdev(), n_bins)
F = dw1.f1(x)

plt.plot(x,F)
plt.axvline(dw1.stdev(), linestyle="dashed", color="black")
plt.ylim(-10,20)
```
