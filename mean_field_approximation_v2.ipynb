```python
import numpy as np
import matplotlib.pyplot as plt
import importlib
import mean_field_approximation_v1 as mfa
```

```python
#MTD parameters
omega = 0.1 #we really care about the limit as this goes to zero but a small finite value should suffice
deltaT = 1000000

#simulation parameter
T = 1
#k_B = 1 assumed here

gw1 = mfa.gaussian_well_1(2,1)

padfactor = 15
x = np.linspace(-padfactor*gw1.stdev(), padfactor*gw1.stdev(),300)
F = gw1.f1(x)
plt.plot(x,F)
plt.axvline(gw1.stdev(), linestyle="dashed", color="black")
```
